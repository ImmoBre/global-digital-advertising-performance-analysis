### Objective

Before performing any analysis, it is essential to evaluate the quality of the dataset. This notebook examines missing values, duplicate records, data types, and potential inconsistencies to ensure the data is suitable for analysis.

#### Missing Values Assessment

```python
missing = df.isnull().sum()

missing
```

```python
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).sort_values(ascending=False)

missing_percentage
```

#### Check for Negative Values

Advertising metrics such as impressions, clicks, spend, conversions, and revenue should not contain negative values.

```python
numeric_cols = [
    "impressions",
    "clicks",
    "ad_spend",
    "conversions",
    "revenue"
]

(df[numeric_cols] < 0).sum()
```

#### Validate Click-Through Rate (CTR)

CTR should approximately equal:

CTR = Clicks / Impressions

```python
calculated_ctr = df["clicks"] / df["impressions"]

comparison = pd.DataFrame({
    "Stored_CTR": df["CTR"],
    "Calculated_CTR": calculated_ctr
})

comparison.head()
```

#### Validate Cost Per Click (CPC)

CPC should approximately equal:

CPC = Ad Spend / Clicks

```python
calculated_cpc = df["ad_spend"] / df["clicks"]

comparison = pd.DataFrame({
    "Stored_CPC": df["CPC"],
    "Calculated_CPC": calculated_cpc
})

comparison.head()
```

#### Validate Return on Ad Spend (ROAS)

ROAS should approximately equal:

ROAS = Revenue / Ad Spend

```python
calculated_roas = df["revenue"] / df["ad_spend"]

comparison = pd.DataFrame({
    "Stored_ROAS": df["ROAS"],
    "Calculated_ROAS": calculated_roas
})

comparison.head()
```
