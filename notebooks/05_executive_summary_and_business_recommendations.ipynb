## Objective

This notebook summarizes the key findings from the analysis and translates them into actionable business recommendations. The goal is to support data-driven decision-making for marketing strategy and budget allocation.

# Executive Summary

This project analyzed digital advertising campaign performance across multiple platforms, industries, campaign types, and countries. The analysis focused on evaluating revenue generation, advertising efficiency, profitability, and campaign effectiveness.

Several performance indicators—including ROAS, profit, conversion rate, and profit margin—were examined to identify opportunities for optimizing marketing investments.

# Key Findings

## Platform Performance

- TikTok Ads generated the highest overall profit.
- Google Ads produced strong revenue but lower profit than TikTok.
- Meta Ads generated the lowest total profit among the three platforms.

## Campaign Performance

- Search and Display campaigns achieved the highest average profit margins.
- Shopping campaigns produced comparatively lower profitability.

## Industry Performance

- SaaS achieved the highest ROAS, narrowly exceeding EdTech.
- EdTech achieved the highest average ROAS, indicating superior advertising efficiency.

## Geographic Performance

- The UAE generated the highest total revenue, while India achieved the highest ROAS.
- The USA produced substantial revenue but recorded the lowest ROAS among the analyzed countries.

## Time Trends

- Revenue and advertising spend fluctuated throughout the year.
- December recorded the highest revenue and advertising investment.
- July represented the weakest month in terms of revenue generation.

## Feature Engineering

- TikTok Ads generated the highest total profit.
- E-commerce achieved the lowest CPM, indicating relatively inexpensive audience reach.
- Conversion rates were generally consistent across countries.

# Strategic Recommendations

## Recommendation 1

Increase investment in high-performing platforms such as TikTok Ads while continuously monitoring return on investment.

---

## Recommendation 2

Expand Search and Display campaigns, as they demonstrate the highest profit margins.

---

## Recommendation 3

Investigate the performance of Shopping campaigns to identify optimization opportunities.

---

## Recommendation 4

Allocate additional marketing resources toward high-performing markets such as the UAE while reviewing strategies in lower-efficiency markets.

---

## Recommendation 5

Leverage the strong advertising efficiency observed in the EdTech industry to maximize future campaign returns.

---

## Recommendation 6

Monitor seasonal performance trends and consider increasing advertising budgets during historically strong periods such as December.

# Business Impact

The findings from this analysis demonstrate that successful marketing strategies depend not only on increasing advertising spend but also on improving campaign efficiency and profitability.

By reallocating budgets toward higher-performing platforms, industries, and campaign types, organizations can potentially increase revenue while maintaining or reducing advertising costs.
