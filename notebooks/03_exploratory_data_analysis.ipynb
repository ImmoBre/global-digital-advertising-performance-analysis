# Global Digital Advertising Performance Analysis

# 03 - Exploratory Data Analysis

## Objective

The purpose of this notebook is to explore campaign performance across platforms, industries, campaign types, and countries. The analysis aims to identify trends, patterns, and business opportunities through descriptive statistics and data visualization.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("ggplot")

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
df = pd.read_csv("../data/global_ads_performance_dataset.csv")

df.head()
```

## Business Question 1

### Which advertising platform generates the highest total revenue?

```python
revenue_by_platform = (
    df.groupby("platform")["revenue"]
    .sum()
    .sort_values(ascending=False)
)

revenue_by_platform
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    revenue_by_platform.index,
    revenue_by_platform.values
)

plt.title("Total Revenue by Platform")
plt.xlabel("Platform")
plt.ylabel("Revenue")

plt.show()
```

### Insight

This visualization compares the total revenue generated by each advertising platform. The results help identify which platform contributes the most to overall business performance and may guide future budget allocation decisions.

# Time-Series Analysis

This section examines how campaign performance changes over time to identify trends and seasonal patterns.

## Business Question 11

### How does total revenue evolve over time?

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
monthly_revenue = (
    df.groupby(df["date"].dt.to_period("M"))["revenue"]
    .sum()
)

monthly_revenue
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    monthly_revenue.index.astype(str),
    monthly_revenue.values,
    marker="o",
    linewidth=2
)

plt.title(
    "Monthly Revenue Trend",
    fontsize=14,
    weight="bold"
)

plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.grid(alpha=0.3)

plt.show()
```

### Business Insight

Monthly revenue fluctuated throughout 2024 rather than following a consistent upward trend. The highest revenue was recorded in **December (approximately 5.16 million)**, while **July generated the lowest revenue (approximately 3.70 million)**.

The strong performance in April and December may indicate seasonal effects or more effective campaign execution during these periods. In contrast, the decline observed in May and July suggests opportunities for campaign optimization or budget reallocation.

## Business Question 12

### How does advertising spend evolve over time?

```python
monthly_spend = (
    df.groupby(df["date"].dt.to_period("M"))["ad_spend"]
    .sum()
)

monthly_spend
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    monthly_spend.index.astype(str),
    monthly_spend.values,
    marker="o",
    linewidth=2
)

plt.title(
    "Monthly Advertising Spend",
    fontsize=14,
    weight="bold"
)

plt.xlabel("Month")
plt.ylabel("Advertising Spend")

plt.xticks(rotation=45)

plt.grid(alpha=0.3)

plt.show()
```

### Business Insight

Advertising expenditure varied across the year, with the highest investment occurring in **December (approximately 1.10 million)** and the lowest in **July (approximately 0.80 million)**.

Interestingly, months with higher advertising spend generally correspond to higher revenue, suggesting a positive relationship between investment and business performance. However, additional analysis is required to determine whether increased spending consistently leads to proportional revenue growth.

# Relationship Analysis

This section investigates the relationships between key marketing metrics to better understand campaign performance.

## Business Question 13

### Is there a relationship between advertising spend and revenue?

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df["ad_spend"],
    df["revenue"],
    alpha=0.6
)

plt.title(
    "Advertising Spend vs Revenue",
    fontsize=14,
    weight="bold"
)

plt.xlabel("Advertising Spend")
plt.ylabel("Revenue")

plt.grid(alpha=0.3)

plt.show()
```

### Business Insight

The scatter plot indicates a generally positive relationship between advertising spend and revenue, suggesting that higher investment tends to generate higher returns.

However, the relationship is not perfectly linear. Several campaigns with substantial advertising spend produced relatively modest revenue, while some lower-spend campaigns achieved strong results. This suggests that campaign effectiveness depends not only on budget size but also on targeting, platform selection, and campaign strategy.

Marketing managers should therefore optimize spending efficiency rather than simply increasing budgets.

## Business Question 14

### Is there a relationship between clicks and conversions?

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df["clicks"],
    df["conversions"],
    alpha=0.6
)

plt.title(
    "Clicks vs Conversions",
    fontsize=14,
    weight="bold"
)

plt.xlabel("Clicks")
plt.ylabel("Conversions")

plt.grid(alpha=0.3)

plt.show()
```

### Business Insight

The analysis reveals a strong positive relationship between clicks and conversions. Campaigns that attract more user engagement generally produce a higher number of conversions.

This finding highlights the importance of improving click generation through effective creatives and targeting strategies, as increased engagement is strongly associated with improved campaign outcomes.

## Business Question 15

### Which variables are most strongly correlated?

```python
correlation = df.select_dtypes(include="number").corr()

correlation
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,8))

plt.imshow(correlation, cmap="Blues", aspect="auto")

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title(
    "Correlation Matrix",
    fontsize=14,
    weight="bold"
)

plt.show()
```

### Business Insight

The correlation matrix demonstrates strong positive relationships among several marketing metrics. In particular:

- Clicks and conversions exhibit a high positive correlation.
- Advertising spend shows a positive relationship with revenue.
- Revenue and ROAS are positively associated.
- CPA displays weaker or negative relationships with several performance metrics, indicating that lower acquisition costs may contribute to improved campaign efficiency.

Overall, the analysis suggests that engagement metrics and investment levels are key drivers of campaign performance.
